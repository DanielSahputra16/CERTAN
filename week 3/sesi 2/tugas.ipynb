```python
import pandas as pd

df = pd.read_csv('C:\\PERKULIHAAN\\SEMESTER 5\\CERTAN\\CERTAN\\week 3\\sesi 2\\rounded_hours_student_scores.csv')
print(df.head())
print(df.columns)
```

```python
X = df[['Hours']].values
y = df['Scores'].values
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y, color='red')
plt.xlabel('Study Time')
plt.ylabel('Score')
plt.title('Study Time vs Score')
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)
```

```python
from sklearn.linear_model import LinearRegression

regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
a = regressor.intercept_
b = regressor.coef_[0]

print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
y_pred = regressor.predict(X_test)

print("Hasil prediksi:")
print(y_pred)
```

```python
from sklearn.metrics import r2_score

score = r2_score(y_test, y_pred)

print("R-Square:", score)
```

```python
plt.scatter(X_train, y_train, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')

plt.title('Score vs Study Time (Training Set)')
plt.xlabel('Study Time')
plt.ylabel('Score')
plt.show()
```

```python
plt.scatter(X_test, y_test, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')

plt.title('Score vs Study Time (Test Set)')
plt.xlabel('Study Time')
plt.ylabel('Score')
plt.show()
```

```python
import pandas as pd

df2 = pd.read_csv('C:\\PERKULIHAAN\\SEMESTER 5\\CERTAN\\CERTAN\\week 3\\sesi 2\\Marketing_Data.csv')
print(df2.head())
print(df2.columns)
```

```python
X2 = df2[['youtube', 'facebook', 'newspaper']]
y2 = df2['sales']
```

```python
from sklearn.model_selection import train_test_split

X_train2, X_test2, y_train2, y_test2 = train_test_split(
    X2, y2, test_size=0.2, random_state=0
)
```

```python
from sklearn.linear_model import LinearRegression

regressor2 = LinearRegression()

regressor2.fit(X_train2, y_train2)
```

```python
y_pred2 = regressor2.predict(X_test2)

print("Hasil prediksi:")
print(y_pred2)
```

```python
a2 = regressor2.intercept_

b1 = regressor2.coef_[0]
b2 = regressor2.coef_[1]
b3 = regressor2.coef_[2]

print("Estimated model intercept, a:", a2)
print("Estimated model slope, b1:", b1)
print("Estimated model slope, b2:", b2)
print("Estimated model slope, b3:", b3)

print(f"ŷ = {a2:.2f} + {b1:.2f}x1 + {b2:.2f}x2 + {b3:.2f}x3")
```

```python
from sklearn.metrics import r2_score

score2 = r2_score(y_test2, y_pred2)

print("R-Square:", score2)
```
