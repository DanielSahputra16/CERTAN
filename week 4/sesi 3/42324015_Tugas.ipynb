Pada bagian ini saya memasukkan library yang dibutuhkan, yaitu NumPy, Pandas, dan dataset Iris.
Setelah dataset dipanggil, datanya dibuat menjadi DataFrame dan ditambahkan kolom Species .
Kemudian x digunakan untuk menyimpan fitur, sedangkan y digunakan untuk menyimpan species
yang nantinya akan menjadi target prediksi.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# Load dataset Iris dari scikit-Learn
iris = load_iris()
```

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names [iris.target]

# Definisikan X (fitur) dan y (target/Label)
X = df.drop(columns=["Species"])
y = df["Species"]

print(X)
```

```python
y
```

Pada bagian ini nama species yang masih berupa teks diubah menjadi angka menggunakan
LabelEncoder . Perubahan ini dilakukan supaya data target bisa diproses oleh model, dan y
digunakan untuk melihat hasil perubahan tersebut.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

Pada bagian ini data dibagi menjadi data training dan data testing. Sebanyak 20% data digunakan
untuk testing, sedangkan sisanya digunakan untuk training. random_state=0 digunakan supaya hasil
pembagian datanya tetap sama ketika kode dijalankan kembali.

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Pada bagian ini saya menggunakan GaussianNB untuk membuat model Naïve Bayes. Setelah itu model
dilatih menggunakan x_train dan y_train supaya model bisa mempelajari pola dari data training
yang diberikan.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

```python
result = NB_model.predict(X_test)
result
```

Pada bagian ini model yang sudah dilatih digunakan untuk memprediksi data testing. Hasil prediksinya
disimpan di result dan ditampilkan untuk melihat species yang diprediksi oleh model dalam bentuk
kode angka.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

Pada bagian ini saya mencoba memberikan data baru ke model untuk melihat species yang diprediksi.
Data baru dimasukkan ke input_data , lalu predict() digunakan untuk melakukan prediksi. Karena
hasilnya masih berupa angka, hasil tersebut diubah kembali menjadi nama species menggunakan
inverse_transform(). Pada contoh di modul, hasilnya adalah setosa.

Warning ini muncul karena data baru yang digunakan sebelumnya tidak mempunyai nama kolom,
sedangkan model dilatih menggunakan DataFrame yang mempunyai nama kolom. Jadi, supaya format
datanya sama, data baru dibuat menjadi DataFrame dengan nama kolom yang sama seperti fitur pada
dataset Iris.

```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab. inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Pada bagian tugas, saya mengganti data contoh sebelumnya dengan data [7.3, 3.8, 4.6, 2.3]. Data
tersebut dimasukkan ke DataFrame dengan nama kolom yang sama seperti dataset Iris, kemudian
diprediksi menggunakan model yang sudah dibuat. Hasil prediksi yang awalnya berupa angka diubah
kembali menjadi nama species, dan hasil akhirnya adalah virginica.

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)

result = NB_model.predict(input_data)
prediction_label = lab. inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```
