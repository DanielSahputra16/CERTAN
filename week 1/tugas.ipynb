```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

# Membuat matriks 5x5 dengan elemen random (bilangan bulat 1-100)
# np.random.seed digunakan supaya hasil random selalu sama tiap dijalankan (opsional, boleh dihapus)
np.random.seed(42)
matriks = np.random.randint(1, 100, size=(5, 5))

print("Matriks 5x5 yang dikonstruksi:")
print(matriks)

print("\nNilai maksimum :", matriks.max())
print("Nilai minimum  :", matriks.min())
print("Nilai rata-rata:", matriks.mean())
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 500)
y = np.linspace(-2, 2, 500)
X, Y = np.meshgrid(x, y)

Z = (
    X**6 + 3*X**4*Y**2 - 3*X**4 + 3*X**2*Y**4 
    - X**2*Y**3 - 6*X**2*Y**2 + 3*X**2 
    + Y**6 - 3*Y**4 + 3*Y**2
)

plt.figure(figsize=(8, 8))

# Hanya gambar kontur tanpa label
plt.contour(X, Y, Z, levels=[1], colors="Blue")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Grafik 2 Dimensi (Tanpa Label)")
plt.grid(True)
plt.axis("equal")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

# Mendefinisikan fungsi z = f(x, y)
def f(x, y):
    return (5*y/4 - np.abs(x))**2 + x**2

x = np.linspace(-2, 2, 100)
y = np.linspace(-2, 2, 100)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Versi 1: surface plot (permukaan)
fig = plt.figure(figsize=(10, 8))
ax = plt.axes(projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('Grafik 3 Dimensi (Surface Plot)')
plt.show()
```
